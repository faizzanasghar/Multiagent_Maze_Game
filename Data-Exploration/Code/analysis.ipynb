# Maze Generation Algorithm Analysis
This notebook analyzes the metrics collected from DFS, BFS, and Prim's algorithms.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('Dataset/maze_metrics.csv')

# Summary Statistics
summary = df.groupby('Algorithm')[['Dead Ends', 'Path Length']].mean()
print(summary)

# Plotting Dead Ends
plt.figure(figsize=(10, 6))
sns.boxplot(x='Algorithm', y='Dead Ends', data=df)
plt.title('Dead Ends by Algorithm')
plt.show()

# Plotting Path Length
plt.figure(figsize=(10, 6))
sns.boxplot(x='Algorithm', y='Path Length', data=df)
plt.title('Path Length by Algorithm')
plt.show()
```
